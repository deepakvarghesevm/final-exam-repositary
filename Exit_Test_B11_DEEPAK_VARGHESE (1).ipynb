```python
# 1. Inspect the unique currencies present in the dataset
unique_currencies = df['revenue_currency'].dropna().unique()
print("Unique currencies in dataset:", unique_currencies)

# 2. Complete exchange rates to USD (approximate rates for normalization)
# 1 Unit of Currency to USD
exchange_rates_to_usd = {
    'USD': 1.0,
    'EUR': 1.08,
    'GBP': 1.26,
    'CAD': 0.74,
    'JPY': 0.0066,
    'INR': 0.012,
    'AUD': 0.65,
    'CHF': 1.11,
    'ZAR': 0.053,
    'SEK': 0.095,
    'HUF': 0.0027,
    'DKK': 0.14,
    'CNY': 0.14,
    'PLN': 0.25,
    'COP': 0.00025,
    'CZK': 0.043,
    'ILS': 0.27,
    'NOK': 0.094,
    'MXN': 0.059,
    'SGD': 0.74,
    'TRY': 0.031,
    'MYR': 0.21,
    'HKD': 0.13,
    'KRW': 0.00075,
    'SAR': 0.27,
    'RUB': 0.011,
    'TWD': 0.031,
    'AED': 0.27,
    'RON': 0.22,
    'KWD': 3.25,
    'THB': 0.028,
    'QAR': 0.27,
    'PEN': 0.27,
    'BRL': 0.20,
    'PHP': 0.018
}

# 3. Create a helper function to convert revenue to USD
def convert_to_usd(row):
    currency = row['revenue_currency']
    val = row['revenue']
    if pd.isna(val) or pd.isna(currency):
        return np.nan
    rate = exchange_rates_to_usd.get(currency.upper(), 1.0)
    return val * rate

# Apply the conversion
df['revenue_usd'] = df.apply(convert_to_usd, axis=1)

# Display top 10 enterprises sorted by standardized USD Revenue
top_usd_enterprises = df.sort_values(by='revenue_usd', ascending=False)[['display_name', 'revenue', 'revenue_currency', 'revenue_usd', 'revenue_year']].head(10)
print("\nTop 10 Enterprises by Correctly Standardized Revenue in USD:")
display(top_usd_enterprises)
```

```python
# 1. Change pandas options to display floats with commas instead of scientific notation
pd.options.display.float_format = '{:,.2f}'.format

# 2. Create a clean helper function to format large numbers to human-readable strings (e.g., $716.92 B)
def format_to_readable_usd(val):
    if pd.isna(val):
        return "N/A"
    if val >= 1e12:
        return f"${val / 1e12:.2f} T" # Trillions
    elif val >= 1e9:
        return f"${val / 1e9:.2f} B"  # Billions
    elif val >= 1e6:
        return f"${val / 1e6:.2f} M"  # Millions
    else:
        return f"${val:,.2f}"

# 3. Create a target formatted column
df['formatted_revenue_usd'] = df['revenue_usd'].apply(format_to_readable_usd)

# Display the top 10 enterprises again with our clean formats
top_formatted = df.sort_values(by='revenue_usd', ascending=False)[['display_name', 'revenue_usd', 'formatted_revenue_usd', 'revenue_year']].head(10)
display(top_formatted)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, r2_score

# 1. Define feature groups from our original cleaned df_drop
# Let's drop the target ('revenue') and identifier columns
features_df = df_drop.drop(columns=[
    'revenue',
    'enterprise_group_id',
    'enterprise_group_name',
    'group_root_lei',
    'display_name',
    'website',
    'website_source',
    'employees_source',
    'revenue_source',
    'net_income_source',
    'formatted_revenue_usd',
    'enterprise_group_entity_status',
    'enterprise_group_jurisdiction',
    'revenue_currency',
    'net_income_currency'
], errors='ignore')

# Identify numerical and categorical columns to preprocess
numeric_features = features_df.select_dtypes(include=[np.number]).columns.tolist()
categorical_features = ['company_type']
binary_features = ['is_standalone_entity']

print("Numeric features to scale:", numeric_features)
print("Categorical features to encode:", categorical_features)

# 2. Build the Column Transformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'), categorical_features)
    ],
    remainder='passthrough' # Keeps 'is_standalone_entity' as-is
)

# 3. Create a pipeline combining preprocessing and modeling
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', DecisionTreeRegressor(random_state=42))
])

# 4. Split original raw target and features (using df_drop)
X_raw = features_df.copy()
y_raw = df_drop['revenue']

X_train_raw, X_test_raw, y_train_raw, y_test_raw = train_test_split(
    X_raw,
    y_raw,
    test_size=0.2,
    random_state=42
)

# 5. Fit the pipeline and predict
pipeline.fit(X_train_raw, y_train_raw)
y_pred_raw = pipeline.predict(X_test_raw)

# 6. Evaluate metrics
print("\nPipeline evaluation:")
print("Mean Absolute Error (MAE):", f"{mean_absolute_error(y_test_raw, y_pred_raw):,.2f}")
print("R-squared (R2) Score:", f"{r2_score(y_test_raw, y_pred_raw):.4f}")
```
